```python
# Create the nested dictionary
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

print("=== ORIGINAL EMPLOYEE DICTIONARY ===")
print("Before:", employee)
print("Explanation: This dictionary contains another dictionary inside 'skills'.\n")

# -----------------------------------------
# Print employee name
# -----------------------------------------

print("Employee Name:", employee["name"])
print("Explanation: Access using employee['name'].\n")

# -----------------------------------------
# Print department
# -----------------------------------------

print("Department:", employee["department"])
print("Explanation: Access using employee['department'].\n")

# -----------------------------------------
# Print complete skills dictionary
# -----------------------------------------

print("Skills Dictionary:", employee["skills"])
print("Explanation: 'skills' itself is a dictionary inside the main dictionary.\n")

# -----------------------------------------
# Print programming language
# -----------------------------------------

print("Programming Language:", employee["skills"]["language"])
print("Explanation: Access nested value using employee['skills']['language'].\n")

# -----------------------------------------
# Print database
# -----------------------------------------

print("Database:", employee["skills"]["database"])
print("Explanation: Access nested value using employee['skills']['database'].\n")

# -----------------------------------------
# Print cloud technology
# -----------------------------------------

print("Cloud Technology:", employee["skills"]["cloud"])
print("Explanation: Access nested value using employee['skills']['cloud'].\n")

# -----------------------------------------
# Change "Python" to "Python + JavaScript"
# -----------------------------------------

print("Before language change:", employee["skills"]["language"])
employee["skills"]["language"] = "Python + JavaScript"
print("After language change:", employee["skills"]["language"])
print("Explanation: Updating a nested dictionary value.\n")

# -----------------------------------------
# Change salary
# -----------------------------------------

print("Before salary change:", employee["salary"])
employee["salary"] = 90000
print("After salary change:", employee["salary"])
print("Explanation: Updating a normal dictionary key.\n")

# -----------------------------------------
# Add "experience": 3
# -----------------------------------------

print("Before adding experience:", employee)
employee["experience"] = 3
print("After adding experience:", employee)
print("Explanation: Adding a new key-value pair to the dictionary.\n")

# -----------------------------------------
# Add another skill under skills dictionary
# -----------------------------------------

print("Before adding new skill:", employee["skills"])
employee["skills"]["devops"] = "Docker"
print("After adding new skill:", employee["skills"])
print("Explanation: Adding a new key-value pair inside the nested dictionary.\n")
```

```python
# Create Student Mark list
marks = [78, 85, 90, 67, 88, 92, 76]

print("=== ORIGINAL MARKS LIST ===")
print("Before:", marks)
print("Explanation: This list stores marks of different subjects.\n")

# -----------------------------------------
# Print the complete list
# -----------------------------------------

print("Complete List:", marks)
print("Explanation: Printing the entire list.\n")

# -----------------------------------------
# Print the first element
# -----------------------------------------

print("First Element:", marks[0])
print("Explanation: marks[0] gives the first value.\n")

# -----------------------------------------
# Print the last element
# -----------------------------------------

print("Last Element:", marks[-1])
print("Explanation: marks[-1] gives the last value.\n")

# -----------------------------------------
# Print elements from index 2 to 5
# -----------------------------------------

print("Slice (index 2 to 5):", marks[2:6])
print("Explanation: marks[2:6] returns elements at index 2,3,4,5.\n")

# -----------------------------------------
# Find the number of elements
# -----------------------------------------

print("Number of Elements:", len(marks))
print("Explanation: len() counts how many items are in the list.\n")

# -----------------------------------------
# Find maximum marks
# -----------------------------------------

print("Maximum Marks:", max(marks))
print("Explanation: max() returns the highest value.\n")

# -----------------------------------------
# Find minimum marks
# -----------------------------------------

print("Minimum Marks:", min(marks))
print("Explanation: min() returns the lowest value.\n")

# -----------------------------------------
# Find total marks
# -----------------------------------------

print("Total Marks:", sum(marks))
print("Explanation: sum() adds all values in the list.\n")

# -----------------------------------------
# Sort marks in ascending order
# -----------------------------------------

print("Before ascending sort:", marks)
marks.sort()
print("After ascending sort:", marks)
print("Explanation: sort() arranges the list in increasing order.\n")

# -----------------------------------------
# Sort marks in descending order
# -----------------------------------------

print("Before descending sort:", marks)
marks.sort(reverse=True)
print("After descending sort:", marks)
print("Explanation: sort(reverse=True) arranges the list in decreasing order.\n")

# -----------------------------------------
# Add 95
# -----------------------------------------

print("Before adding 95:", marks)
marks.append(95)
print("After adding 95:", marks)
print("Explanation: append() adds one item at the end.\n")

# -----------------------------------------
# Add [81, 84]
# -----------------------------------------

print("Before extend():", marks)
marks.extend([81, 84])
print("After extend():", marks)
print("Explanation: extend() adds multiple items individually.\n")

# -----------------------------------------
# Remove 67
# -----------------------------------------

print("Before removing 67:", marks)
marks.remove(67)
print("After removing 67:", marks)
print("Explanation: remove() deletes the first matching value.\n")

# -----------------------------------------
# Count how many times 90 occurs
# -----------------------------------------

print("Count of 90:", marks.count(90))
print("Explanation: count() tells how many times 90 appears.\n")

# -----------------------------------------
# Find the index of 88
# -----------------------------------------

print("Index of 88:", marks.index(88))
print("Explanation: index() returns the position of 88.\n")
```

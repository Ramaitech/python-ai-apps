```python
# Create a Mini E-Commerce Dataset
# Create the list of dictionaries
products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]

print("=== ORIGINAL PRODUCTS DATASET ===")
print("Before:", products)
print("Explanation: This is a list containing three dictionaries.\n")

# -----------------------------------------
# Print all products
# -----------------------------------------

print("All Products:", products)
print("Explanation: Printing the entire list of dictionaries.\n")

# -----------------------------------------
# Print first product
# -----------------------------------------

print("First Product:", products[0])
print("Explanation: products[0] gives the first dictionary.\n")

# -----------------------------------------
# Print second product's price
# -----------------------------------------

print("Second Product Price:", products[1]["price"])
print("Explanation: products[1] selects second dictionary, ['price'] selects its price.\n")

# -----------------------------------------
# Print third product's brand
# -----------------------------------------

print("Third Product Brand:", products[2]["brand"])
print("Explanation: products[2] selects third dictionary, ['brand'] selects its brand.\n")

# -----------------------------------------
# Change first product's price
# -----------------------------------------

print("Before price change:", products[0]["price"])
products[0]["price"] = 75000
print("After price change:", products[0]["price"])
print("Explanation: Updating a value inside the first dictionary.\n")

# -----------------------------------------
# Add "rating" to the second product
# -----------------------------------------

print("Before adding rating:", products[1])
products[1]["rating"] = 4.5
print("After adding rating:", products[1])
print("Explanation: Adding a new key-value pair to the second dictionary.\n")

# -----------------------------------------
# Add another product manually
# -----------------------------------------

print("Before adding new product:", products)
products.append({
    "name": "Smartwatch",
    "price": 15000,
    "brand": "Fitbit"
})
print("After adding new product:", products)
print("Explanation: append() adds a new dictionary to the list.\n")

# -----------------------------------------
# Print final dataset
# -----------------------------------------

print("=== FINAL PRODUCTS DATASET ===")
print("After:", products)
print("Explanation: This shows all products after modifications.\n")
```

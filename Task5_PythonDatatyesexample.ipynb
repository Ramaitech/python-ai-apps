```python
"""
Demonstration of Python data types, type identification,
and basic type conversions with explanations.
"""
# -----------------------------------------
# Creating variables of different data types
# -----------------------------------------

# int examples
a = 10
b = -5

# float examples
c = 3.14
d = 0.001

# string examples
e = "Hello"
f = "123"

# boolean examples
g = True
h = False

# list examples
i = [1, 2, 3]
j = ["apple", "banana", "cherry"]

# tuple examples
k = (10, 20, 30)
l = ("red", "green", "blue")

# set examples
m = {1, 2, 3}
n = {"A", "B", "C"}

# dictionary examples
o = {"name": "John", "age": 25}
p = {"brand": "Toyota", "year": 2024}

# -----------------------------------------
# Printing data types
# -----------------------------------------

print("=== DATA TYPES OF VARIABLES ===")
print(a, "=>", type(a))
print(b, "=>", type(b))
print(c, "=>", type(c))
print(d, "=>", type(d))
print(e, "=>", type(e))
print(f, "=>", type(f))
print(g, "=>", type(g))
print(h, "=>", type(h))
print(i, "=>", type(i))
print(j, "=>", type(j))
print(k, "=>", type(k))
print(l, "=>", type(l))
print(m, "=>", type(m))
print(n, "=>", type(n))
print(o, "=>", type(o))
print(p, "=>", type(p), "\n")

# -----------------------------------------
# Type Conversion Examples (Before + After)
# -----------------------------------------

# 1. int("100")
print("Before int():", "100", "=>", type("100"))
converted_int = int("100")
print("After int():", converted_int, "=>", type(converted_int))
print("Explanation: The string '100' becomes an integer.\n")

# 2. float("45.67")
print("Before float():", "45.67", "=>", type("45.67"))
converted_float = float("45.67")
print("After float():", converted_float, "=>", type(converted_float))
print("Explanation: The string '45.67' becomes a floating-point number.\n")

# 3. str(500)
print("Before str():", 500, "=>", type(500))
converted_str = str(500)
print("After str():", converted_str, "=>", type(converted_str))
print("Explanation: The number 500 becomes a string.\n")

# 4. bool(1)
print("Before bool():", 1, "=>", type(1))
converted_bool = bool(1)
print("After bool():", converted_bool, "=>", type(converted_bool))
print("Explanation: Any non-zero number becomes True.\n")

# 5. list((1, 2, 3))
print("Before list():", (1, 2, 3), "=>", type((1, 2, 3)))
converted_list = list((1, 2, 3))
print("After list():", converted_list, "=>", type(converted_list))
print("Explanation: Tuple becomes a list (mutable).\n")

# 6. tuple([1, 2, 3])
print("Before tuple():", [1, 2, 3], "=>", type([1, 2, 3]))
converted_tuple = tuple([1, 2, 3])
print("After tuple():", converted_tuple, "=>", type(converted_tuple))
print("Explanation: List becomes a tuple (immutable).\n")

# 7. set([1, 2, 2, 3])
print("Before set():", [1, 2, 2, 3], "=>", type([1, 2, 2, 3]))
converted_set = set([1, 2, 2, 3])
print("After set():", converted_set, "=>", type(converted_set))
print("Explanation: Set removes duplicate values automatically.\n")
```

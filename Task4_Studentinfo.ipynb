```python
# Student Information challenge
# Student string operations
student = "python programming for data science"

#  String slicing operations
# 1. Print the complete string
print("Complete string:", student)

# 2.Print the first character
first_character = student[0]

# 3. Print the last character
last_character = student[-1]

# 4. Print the first six characters
first_six_characters = student[0:6]

# 5. Print the last seven characters
last_seven_characters = student[-7:]

# 6. Reverse the string using slicing
reversed_string = student[::-1]

# 7. Print the string in uppercase
uppercase_string = student.upper()

# 8.Convert it to lowercase
lowercase_string = student.lower()

# 9.Convert it to title case
title_case_string = student.title()

# 10.Count how many times "a" appears
count_a = student.count("a")

# 11. Find the position of "programming"
position_programming = student.find("programming")

# 12. Replace "data science" with "artificial intelligence"
replaced_string = student.replace("data science", "artificial intelligence")

# 13.Split the string into individual words
split_words = student.split()

# Display the results all the string operations
print("First character:", first_character)  
print("Last character:", last_character)
print("First six characters:", first_six_characters)
print("Last seven characters:", last_seven_characters)
print("Reversed string:", reversed_string)
print("Uppercase string:", uppercase_string)
print("Lowercase string:", lowercase_string)
print("Title case string:", title_case_string)
print("Count of 'a':", count_a)
print("Position of 'programming':", position_programming)
print("Replaced string:", replaced_string)
print("Split words:", split_words)

```

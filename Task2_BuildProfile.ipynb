```python
#Build Your Developer Profile Using Variables
name=input("What is your name? ")
age=int(input("What is your age? "))
college=input("What college did you attend? ")
company=input("What company do you work for? ")
current_learningtopic=input("What are you currently learning? ")
years_exp=int(input("How many years of experience do you have? "))
like_python=bool(input("Do you like Python? (yes/no) "))
expect_salary=float(input("Your expected salary after learning Python? "))

#Display the developer's profile
print("Developer Profile:")
print("Name:", name)
print("Age:", age)
print("College:", college)
print("Company:", company)
print("Currently Learning:", current_learningtopic)
print("Years of Experience:", years_exp)
print("Likes Python:", like_python)
print("Expected Salary:", expect_salary)

#Print the data type of each variable   
print("Data Types:")
print("Name:", type(name))
print("Age:", type(age))
print("College:", type(college))
print("Company:", type(company))
print("Currently Learning:", type(current_learningtopic))
print("Years of Experience:", type(years_exp))
print("Likes Python:", type(like_python))
print("Expected Salary:", type(expect_salary))      
```

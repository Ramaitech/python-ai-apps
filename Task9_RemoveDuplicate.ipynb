```python
# Original list

# Remove Duplicate data using set 
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

print("=== ORIGINAL LIST ===")
print("Before:", numbers)
print("Explanation: This list contains duplicate numbers.\n")

# Convert list to set
num_set = set(numbers)
print("=== CONVERT LIST TO SET ===")
print("Before:", numbers)
print("After:", num_set)
print("Explanation: Sets automatically remove duplicates.\n")

# Observe duplicates disappearing
print("Original Count:", len(numbers))
print("Unique Count:", len(num_set))
print("Explanation: The number of items reduces because duplicates are removed.\n")

# Convert set back to list
new_list = list(num_set)
print("=== CONVERT SET BACK TO LIST ===")
print("Before:", num_set)
print("After:", new_list)
print("Explanation: Converting back to a list gives unique values only.\n")

# Number of original elements
print("Number of Original Elements:", len(numbers))
print("Explanation: Count of items before removing duplicates.\n")

# Number of unique elements
print("Number of Unique Elements:", len(num_set))
print("Explanation: Count of items after removing duplicates.\n")

# Another example with duplicate student names
students = ["Rahul", "Priya", "Rahul", "Aman", "Priya", "Karan"]

print("=== STUDENT LIST WITH DUPLICATES ===")
print("Before:", students)
student_set = set(students)
print("After:", student_set)
print("Explanation: Duplicate student names are removed when converted to a set.\n")

# Why sets are useful
print("=== WHY SETS ARE USEFUL ===")
print("Sets are useful because they automatically remove duplicates.")
print("They help clean data, find unique values, and avoid repetition.\n")
```

```python
# Create student dictionary

student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

print("=== ORIGINAL STUDENT DICTIONARY ===")
print("Before:", student)
print("Explanation: This dictionary stores student details.\n")

# -----------------------------------------
# Print the complete dictionary
# -----------------------------------------

print("Complete Dictionary:", student)
print("Explanation: Printing the entire dictionary.\n")

# -----------------------------------------
# Print the student's name
# -----------------------------------------

print("Student Name:", student["name"])
print("Explanation: Access using student['name'].\n")

# -----------------------------------------
# Print their course
# -----------------------------------------

print("Course:", student["course"])
print("Explanation: Access using student['course'].\n")

# -----------------------------------------
# Print all keys
# -----------------------------------------

print("Keys:", student.keys())
print("Explanation: keys() returns all dictionary keys.\n")

# -----------------------------------------
# Print all values
# -----------------------------------------

print("Values:", student.values())
print("Explanation: values() returns all dictionary values.\n")

# -----------------------------------------
# Print all key-value pairs
# -----------------------------------------

print("Items:", student.items())
print("Explanation: items() returns key-value pairs.\n")

# -----------------------------------------
# Change marks from 88 to 92
# -----------------------------------------

print("Before marks change:", student["marks"])
student["marks"] = 92
print("After marks change:", student["marks"])
print("Explanation: Updating an existing key-value pair.\n")

# -----------------------------------------
# Add "email"
# -----------------------------------------

print("Before adding email:", student)
student["email"] = "rahul@example.com"
print("After adding email:", student)
print("Explanation: Adding a new key-value pair.\n")

# -----------------------------------------
# Add "phone"
# -----------------------------------------

print("Before adding phone:", student)
student["phone"] = "9876543210"
print("After adding phone:", student)
print("Explanation: Adding another new key-value pair.\n")

# -----------------------------------------
# Remove "city"
# -----------------------------------------

print("Before removing city:", student)
student.pop("city")
print("After removing city:", student)
print("Explanation: pop('city') removes the key-value pair.\n")

# -----------------------------------------
# Use get() to retrieve "name"
# -----------------------------------------

print("Using get() for name:", student.get("name"))
print("Explanation: get() safely retrieves a value without error.\n")

# -----------------------------------------
# Create a copy of the dictionary
# -----------------------------------------

print("Before copy():", student)
student_copy = student.copy()
print("After copy():", student_copy)
print("Explanation: copy() creates a separate duplicate dictionary.\n")
```

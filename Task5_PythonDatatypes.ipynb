```python
"""
Demonstration of Python data types, type identification,
and basic type conversions with explanations.
"""
print("=== TYPE CONVERSION WITH ORIGINAL & NEW DATA TYPES ===\n")

# 1. string → int
original = "100"
converted = int(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: Text '100' becomes number 100.\n")

# 2. string → float
original = "45.67"
converted = float(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: Text '45.67' becomes decimal number 45.67.\n")

# 3. int → string
original = 500
converted = str(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: Number 500 becomes text '500'.\n")

# 4. int → bool
original = 1
converted = bool(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: 1 becomes True because any non-zero number is True.\n")

# 5. tuple → list
original = (1, 2, 3)
converted = list(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: Tuple becomes a list so it can be changed.\n")

# 6. list → tuple
original = [1, 2, 3]
converted = tuple(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: List becomes a tuple which cannot be changed.\n")

# 7. list → set
original = [1, 2, 2, 3]
converted = set(original)
print("Original:", original, "=>", type(original))
print("Converted:", converted, "=>", type(converted))
print("Explanation: Set removes duplicates, so only unique values remain.\n")
```

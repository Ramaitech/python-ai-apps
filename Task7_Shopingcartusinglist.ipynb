```python
# Shopping Cart Using a Python List

# Initial cart list
# Initial cart list
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

print("=== ORIGINAL CART ===")
print("Before:", cart)
print("Explanation: This is the starting list of products.\n")

# -----------------------------------------
# Display all products
# -----------------------------------------

print("All Products:", cart)
print("Explanation: Printing the entire list.\n")

# -----------------------------------------
# Access first and last products
# -----------------------------------------

print("First Product:", cart[0])
print("Explanation: cart[0] gives the first item.\n")

print("Last Product:", cart[-1])
print("Explanation: cart[-1] gives the last item.\n")

# -----------------------------------------
# Add "Webcam" using append()
# -----------------------------------------

print("Before append():", cart)
cart.append("Webcam")
print("After append():", cart)
print("Explanation: append() adds ONE item at the end of the list.\n")

# -----------------------------------------
# Insert "USB Hub" at index 2
# -----------------------------------------

print("Before insert():", cart)
cart.insert(2, "USB Hub")
print("After insert():", cart)
print("Explanation: insert(index, item) places the item at the given position.\n")

# -----------------------------------------
# Remove "Mouse"
# -----------------------------------------

print("Before remove():", cart)
cart.remove("Mouse")
print("After remove():", cart)
print("Explanation: remove('Mouse') deletes the FIRST matching item.\n")

# -----------------------------------------
# Remove last item using pop()
# -----------------------------------------

print("Before pop():", cart)
removed_item = cart.pop()
print("Removed Item:", removed_item)
print("After pop():", cart)
print("Explanation: pop() removes and RETURNS the last item.\n")

# -----------------------------------------
# Find index of "Monitor"
# -----------------------------------------

print("Index of Monitor:", cart.index("Monitor"))
print("Explanation: index('Monitor') tells the position of Monitor.\n")

# -----------------------------------------
# Count occurrences of "Laptop"
# -----------------------------------------

print("Laptop Count:", cart.count("Laptop"))
print("Explanation: count('Laptop') shows how many times Laptop appears.\n")

# -----------------------------------------
# Create a copy of the cart
# -----------------------------------------

print("Before copy():", cart)
cart_copy = cart.copy()
print("After copy():", cart_copy)
print("Explanation: copy() creates a separate duplicate list.\n")

# -----------------------------------------
# Reverse the cart
# -----------------------------------------

print("Before reverse():", cart)
cart.reverse()
print("After reverse():", cart)
print("Explanation: reverse() flips the list order.\n")

# -----------------------------------------
# Sort the products alphabetically
# -----------------------------------------

print("Before sort():", cart)
cart.sort()
print("After sort():", cart)
print("Explanation: sort() arranges items alphabetically.\n")

# -----------------------------------------
# Demonstrate extend()
# -----------------------------------------

print("Before extend():", cart)
cart.extend(["Charger", "USB Cable"])
print("After extend():", cart)
print("Explanation: extend() adds MULTIPLE items individually to the list.\n")

# -----------------------------------------
# Demonstrate clear()
# -----------------------------------------

print("Before clear():", cart)
cart.clear()
print("After clear():", cart)
print("Explanation: clear() removes ALL items, leaving an empty list.\n")
```

```python
# Nested List Challenge

# -----------------------------------------
# Original nested list
# -----------------------------------------

students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]

print("=== ORIGINAL STUDENT LIST ===")
print("Before:", students)
print("Explanation: This is a nested list where each inner list is a student record.\n")

# -----------------------------------------
# Print the complete list
# -----------------------------------------

print("Complete List:", students)
print("Explanation: Printing the entire nested list.\n")

# -----------------------------------------
# Print Rahul's name
# -----------------------------------------

print("Rahul's Name:", students[0][0])
print("Explanation: students[0] is Rahul's record; students[0][0] is his name.\n")

# -----------------------------------------
# Print Priya's age
# -----------------------------------------

print("Priya's Age:", students[1][1])
print("Explanation: students[1] is Priya's record; students[1][1] is her age.\n")

# -----------------------------------------
# Print Aman's course
# -----------------------------------------

print("Aman's Course:", students[2][2])
print("Explanation: students[2] is Aman's record; students[2][2] is his course.\n")

# -----------------------------------------
# Print Priya's complete record
# -----------------------------------------

print("Priya's Record:", students[1])
print("Explanation: students[1] gives the entire list for Priya.\n")

# -----------------------------------------
# Change Rahul's course to "AI"
# -----------------------------------------

print("Before Changing Rahul's Course:", students[0])
students[0][2] = "AI"
print("After Changing Rahul's Course:", students[0])
print("Explanation: students[0][2] was 'Python', now updated to 'AI'.\n")

# -----------------------------------------
# Add another student manually
# -----------------------------------------

print("Before Adding New Student:", students)
students.append(["Sneha", 23, "Cyber Security"])
print("After Adding New Student:", students)
print("Explanation: append() adds a new student record at the end.\n")

# -----------------------------------------
# Print updated nested list
# -----------------------------------------

print("=== UPDATED STUDENT LIST ===")
print("After:", students)
print("Explanation: This shows the final list after all changes.\n")
```

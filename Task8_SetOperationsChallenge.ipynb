```python
# Set Operation Challenage
## Create sets
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

print("=== ORIGINAL SETS ===")
print("Python Students:", python_students)
print("Java Students:", java_students)
print("Explanation: Each set contains unique student names.\n")

# -----------------------------------------
# union() → students learning Python OR Java
# -----------------------------------------

print("Before union():", python_students, java_students)
either = python_students.union(java_students)
print("After union():", either)
print("Explanation: union() combines both sets and removes duplicates.\n")

# -----------------------------------------
# intersection() → students learning BOTH
# -----------------------------------------

print("Before intersection():", python_students, java_students)
both = python_students.intersection(java_students)
print("After intersection():", both)
print("Explanation: intersection() gives only the common students.\n")

# -----------------------------------------
# difference() → only Python students
# -----------------------------------------

print("Before difference():", python_students, java_students)
only_python = python_students.difference(java_students)
print("After difference():", only_python)
print("Explanation: difference() removes Java students from Python group.\n")

# -----------------------------------------
# difference() → only Java students
# -----------------------------------------

print("Before difference():", java_students, python_students)
only_java = java_students.difference(python_students)
print("After difference():", only_java)
print("Explanation: difference() removes Python students from Java group.\n")

# -----------------------------------------
# symmetric_difference() → students in EXACTLY one group
# -----------------------------------------

print("Before symmetric_difference():", python_students, java_students)
exactly_one = python_students.symmetric_difference(java_students)
print("After symmetric_difference():", exactly_one)
print("Explanation: symmetric_difference() keeps only non‑common students.\n")

# -----------------------------------------
# add() → add a new student
# -----------------------------------------

print("Before add():", python_students)
python_students.add("Meena")
print("After add():", python_students)
print("Explanation: add() inserts a new student into the set.\n")

# -----------------------------------------
# remove() → remove an existing student
# -----------------------------------------

print("Before remove():", python_students)
python_students.remove("Aman")
print("After remove():", python_students)
print("Explanation: remove() deletes the student; ERROR if not found.\n")

# -----------------------------------------
# discard() → safe remove (no error)
# -----------------------------------------

print("Before discard():", python_students)
python_students.discard("XYZ")   # XYZ does not exist
print("After discard():", python_students)
print("Explanation: discard() does NOT give an error even if the student does not exist.\n")
```

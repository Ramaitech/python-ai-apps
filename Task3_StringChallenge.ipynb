```python
#String Cleaning Challenge
# Original string
message = "   Welcome To Python Programming Class   "

print("=== ORIGINAL STRING ===")
print("Before:", message)
print("Explanation: The string has extra spaces at the beginning and end.\n")

# 1. Remove extra spaces → strip()
before_strip = message
after_strip = message.strip()
print("Before strip():", before_strip)
print("After strip():", after_strip)
print("Explanation: strip() removes spaces from both ends.\n")

# 2. Convert to lowercase → lower()
before_lower = after_strip
after_lower = after_strip.lower()
print("Before lower():", before_lower)
print("After lower():", after_lower)
print("Explanation: lower() converts all characters to lowercase.\n")

# 3. Convert to uppercase → upper()
before_upper = after_strip
after_upper = after_strip.upper()
print("Before upper():", before_upper)
print("After upper():", after_upper)
print("Explanation: upper() converts all characters to uppercase.\n")

# 4. Convert to title case → title()
before_title = after_strip
after_title = after_strip.title()
print("Before title():", before_title)
print("After title():", after_title)
print("Explanation: title() capitalizes the first letter of every word.\n")

# 5. Replace "Python" with "Advanced Python" → replace()
before_replace = after_strip
after_replace = after_strip.replace("Python", "Advanced Python")
print("Before replace():", before_replace)
print("After replace():", after_replace)
print("Explanation: replace() substitutes one substring with another.\n")

# 6. Check whether string starts with "Welcome" → startswith()
print("Does it start with 'Welcome'?:", after_strip.startswith("Welcome"))
print("Explanation: startswith() checks the beginning of the string.\n")

# 7. Check whether string ends with "Class" → endswith()
print("Does it end with 'Class'?:", after_strip.endswith("Class"))
print("Explanation: endswith() checks the ending of the string.\n")

# 8. Count occurrences of "o" → count()
print("Count of 'o':", after_strip.count("o"))
print("Explanation: count() tells how many times a substring appears.\n")

# 9. Find the position of "Programming" → find()
print("Position of 'Programming':", after_strip.find("Programming"))
print("Explanation: find() returns the index of the substring (or -1 if not found).\n")

# 10. Split the sentence into words → split()
before_split = after_strip
after_split = after_strip.split()
print("Before split():", before_split)
print("After split():", after_split)
print("Explanation: split() breaks the string into a list of words.\n")
```

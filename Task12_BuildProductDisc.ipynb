```python
# Create the nested dictionary
# Create the dictionary
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}

print("=== ORIGINAL LAPTOP DICTIONARY ===")
print("Before:", laptop)
print("Explanation: This dictionary represents a laptop with various properties.\n")

# -----------------------------------------
# Display the product
# -----------------------------------------

print("Laptop Product:", laptop)
print("Explanation: Printing the entire dictionary.\n")

# -----------------------------------------
# Print brand
# -----------------------------------------

print("Brand:", laptop["brand"])
print("Explanation: Access using laptop['brand'].\n")

# -----------------------------------------
# Print model
# -----------------------------------------

print("Model:", laptop["model"])
print("Explanation: Access using laptop['model'].\n")

# -----------------------------------------
# Print price
# -----------------------------------------

print("Price:", laptop["price"])
print("Explanation: Access using laptop['price'].\n")

# -----------------------------------------
# Change price
# -----------------------------------------

print("Before price change:", laptop["price"])
laptop["price"] = 110000
print("After price change:", laptop["price"])
print("Explanation: Updating the value of an existing key.\n")

# -----------------------------------------
# Add processor information
# -----------------------------------------

print("Before adding processor:", laptop)
laptop["processor"] = "Intel i7"
print("After adding processor:", laptop)
print("Explanation: Adding a new key-value pair.\n")

# -----------------------------------------
# Add GPU information
# -----------------------------------------

print("Before adding GPU:", laptop)
laptop["gpu"] = "NVIDIA RTX 3050"
print("After adding GPU:", laptop)
print("Explanation: Adding another new key-value pair.\n")

# -----------------------------------------
# Change RAM to "32GB"
# -----------------------------------------

print("Before RAM change:", laptop["ram"])
laptop["ram"] = "32GB"
print("After RAM change:", laptop["ram"])
print("Explanation: Updating an existing key.\n")

# -----------------------------------------
# Remove "available"
# -----------------------------------------

print("Before removing 'available':", laptop)
laptop.pop("available")
print("After removing 'available':", laptop)
print("Explanation: pop() removes a key-value pair.\n")

# -----------------------------------------
# Print all keys
# -----------------------------------------

print("Keys:", laptop.keys())
print("Explanation: keys() returns all dictionary keys.\n")

# -----------------------------------------
# Print all values
# -----------------------------------------

print("Values:", laptop.values())
print("Explanation: values() returns all dictionary values.\n")

# -----------------------------------------
# Print all items
# -----------------------------------------

print("Items:", laptop.items())
print("Explanation: items() returns key-value pairs.\n")

# -----------------------------------------
# Create second product dictionary (mobile phone)
# -----------------------------------------

mobile = {
    "brand": "Samsung",
    "model": "Galaxy S24",
    "price": 85000,
    "ram": "12GB",
    "storage": "256GB",
    "battery": "5000mAh"
}

print("\n=== SECOND PRODUCT (MOBILE PHONE) ===")
print("Mobile:", mobile)
print("Explanation: This is another dictionary representing a mobile phone.\n")
```
